# Bar Graph in Matplotlib

A bar graph (or bar chart) represents categorical data with rectangular bars.

- The height (or length) of the bar shows the value.
- Categories are usually along the x-axis (horizontal), and values along the y-axis (vertical).

We use bar graphs when we want to compare quantities across categories.

## Simple Vertical Bar Graph


```python
import matplotlib.pyplot as plt

# Data
fruits = ["Apple", "Banana", "Orange"]
sales = [10, 15, 7]
# Plot
bars=plt.bar(fruits, sales, color="skyblue")
# Add values on top
plt.bar_label(bars, label_type="edge") 
plt.xlabel("Fruits")
plt.ylabel("Sales")
plt.title("Fruit Sales")
plt.show()
```

## Simple Horizontal Bar Chart
We can use `barh` function to create a vertical Bar Graph 

```python
bars = plt.barh(fruits, sales, color="lightgreen")

plt.bar_label(bars, label_type="edge")
plt.xlabel("Sales")
plt.ylabel("Fruits")
plt.title("Fruit Sales with Values (Horizontal)")
plt.show()
```

## Grouped Bar Chart

```python
import numpy as np

fruits = ["Apple", "Banana", "Orange"]
sales_2024 = [10, 15, 7]
sales_2025 = [12, 18, 5]

x = np.arange(len(fruits))
width = 0.35

bars1 = plt.bar(x - width/2, sales_2024, width, label="2024")
bars2 = plt.bar(x + width/2, sales_2025, width, label="2025")

plt.bar_label(bars1, padding=1)
plt.bar_label(bars2, padding=1)

plt.xticks(x, fruits)
plt.xlabel("Fruits")
plt.ylabel("Sales")
plt.title("Fruit Sales Comparison with Values")
plt.legend()
plt.show()
```
