```python
import numpy as np

fruits = ["Apple", "Banana", "Orange"]
sales_2024 = [10, 15, 7]
sales_2025 = [12, 18, 5]

x = np.arange(len(fruits))
width = 0.35

bars1 = plt.bar(x - width/2, sales_2024, width, label="2024")
bars2 = plt.bar(x + width/2, sales_2025, width, label="2025")

plt.bar_label(bars1, padding=1)
plt.bar_label(bars2, padding=1)

plt.xticks(x, fruits)
plt.xlabel("Fruits")
plt.ylabel("Sales")
plt.title("Fruit Sales Comparison with Values")
plt.legend()
plt.show()
```

## Horizontal Bar Plot

```python
df = pd.read_csv('data_program.csv')
```

```python
df.head()
```

```python
ids = df['Responder_id']
languages_worked_with = df['LanguagesWorkedWith']
```

```python
from collections import Counter
```

```python
counter = Counter()
```

```python
for language in languages_worked_with:
    counter.update(language.split(';'))
```

```python
counter.most_common(15)
```

```python
languages = []
popularity = []

for item in counter.most_common(15):
    languages.append(item[0])
    popularity.append(item[1])

languages.reverse()
popularity.reverse()
```

```python
plt.barh(languages,popularity)
plt.title('Bar Plot of Popular Programming Language')
```
