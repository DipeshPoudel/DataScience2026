# Matplotlib

[Matplotlib](https://matplotlib.org/) is a plotting library for python. It is a very powerful plotting library useful with Python and NumPy. It was also used in while taking the image of blackhole. The most used module of Matplotib is Pyplot which provides an interface like MATLAB but instead.

Below is the description given by the [NumFocus](https://numfocus.org/) group which maintains the projects that is wildly used in data analysis like pandas, numpy, SciPy, matplotlib and so on. 
Matplotlib is a Python 2D plotting library which produces publication-quality figures in a variety of hardcopy formats and interactive environments across platforms. Matplotlib can be used in Python scripts, the Python and IPython shell, the Jupyter notebook, web application servers, and seven graphical user interface toolkits. For simple plotting, the pyplot module provides a MATLAB-like interface, particularly when combined with IPython. Full control of line styles, font properties, axes properties, etc. is accessible via an object-oriented interface.[NumFocus](https://numfocus.org/project/matplotlib)

## Bar Graph

A bar graph is a graph that shows categorical data with rectangular bars with heights or lengths proportional to the values that they represent. The bars can be plotted vertically or horizontally.

````python
df=df.set_index('State')
````

````python
plt.bar(df.index,df['Generation (GWh)'])
plt.ylabel('Solar Power Generation')
plt.xlabel('State')
plt.title('Bar Graph Showing Power Generation in Various state of US')
plt.show()
````

The example above is of vertical bar graph. We can use `barh` method for horizontal bar graph. We will be also showcasing how we can set the temporary style for a particular graph.

````python
with plt.style.context('Solarize_Light2'):
    plt.barh(df.index,df['Generation (GWh)'])
    plt.xlabel('Solar Power Generation')
    plt.ylabel('State')
    plt.title('Bar Graph Showing Power Generation in Various state of US')
plt.show()
````

We can plot multiple bar graph for same category. These type of bar graph is called grounded bar graph 

````python
labels = df.index
x = np.arange(len(labels))  # the label locations
width = 0.35  # the width of the bars
fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, df['Installed Capacity (MW)'], width, label='Installed Capacity')
rects2 = ax.bar(x + width/2, df['Generation (GWh)'], width, label='Generation')
ax.set_title('Installed and Geneartion')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()
plt.show()
````

### Pie Chart

A pie chart displays data, information, and statistics in an easy-to-read 'pie-slice' format with varying slice sizes telling you how much of one data element exists. The bigger the slice, the more of that particular data was gathered.

````python
fig1, ax1 = plt.subplots()
ax1.pie(df['Number of Solar Plants'],  labels=df.index, autopct='%1.1f%%')
ax1.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.legend()
plt.title('Pie Chart')
plt.show()
````

## Matplotlib in Machine Learning

There is a saying that `picture speaks thousands words` and it is very true in machine learning. Various kind of graph can help us in various stages of building a machine learning model. We will discuss some of the plots that can help us in each stage of building the machine learning model.

### Understanding Data

To build any machine learning data we need to understand the data. Based on the data we can take descisions like what tranformation to perform, what algorithm should we choose and what can be some optimal hyperparameters. For Example we will be taking iris dataset.

````python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import datasets
````

````python
iris = datasets.load_iris()
````

````python
iris_df = pd.DataFrame(iris.data,columns=iris.feature_names)
````

````python
iris_df['target']=iris.target
````

````python
iris_df.head()
````

````python
iris_df['target'].value_counts().plot(kind='bar')
plt.title('Bar Graph Showing Number of Target Classes')
plt.xlabel('Target Class')
plt.ylabel('Frequency')
plt.show()
````

From the above bar graph we can see that there are three target class. There is no class imbalance problem in the dataset. If we had example one class very high and other very few then we would have to think about making the dataset balanced.

In cluster analysis we can look at the data and say how many cluster there could be

````python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
plt.style.use('seaborn') 
X1, Y1 = make_blobs(n_features=4, centers=3,random_state=10)
plt.scatter(X1[:, 0], X1[:, 1], marker='o', c=Y1,s=80,cmap='rainbow')
plt.title("Example Data Which Have Three Clusters")
plt.show()
````

We can also plot the correlation matrix between the variables to show the correlation among the variables.

````python
corr = iris_df.corr()
corr.style.background_gradient(cmap='coolwarm')
````

We can use Histogram to check the distribution of the data. Some of the machine learning algorithm have assumptions
that the data should follow certain kind of distribution. For example in linear rgression data should follow the normal disribution 

````python
plt.hist(iris_df['sepal length (cm)'])
plt.show()
````

### Feature Selection

Feature selection is one of the important steps in machine learning. We can plot the relative importance of the features to see which are the dominant one. It will help us choose the best features from the higher number of features. 

````python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(iris.data,iris.target)
importances = rf.feature_importances_
feat_importances = pd.Series(importances, index=iris.feature_names)
feat_importances.nlargest(6).plot(kind='barh')
plt.show()
````

We can see that the `petal width(cm)` is the most important features. Most of the time the decision to keep and leave features is decided on the basis of domain knowledge.

## Model Evaluation

We can plot various graphs to evaluate a machine learning. We will discuss some of the graphs below

````python
from sklearn import metrics
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
````

````python
X_train,X_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=0)
````

````python
rf=RandomForestClassifier()
````

````python
rf.fit(X_train,y_train)
````

````python
y_pred=rf.predict(X_test)
````

````python
plt.style.use('ggplot')
cm = metrics.confusion_matrix(y_test,y_pred)
metrics.plot_confusion_matrix(rf,X_test,y_test)
plt.title("Confusion Matrix")
plt.show()
````

## Comparision of Models

We can bar graph to show the accuracy of Multiple machine learning model to see which one performed the best. The example below shows 7 classifier being trained. We will plot the bar graph showing accuracy of each model.

````python
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC, LinearSVC, NuSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB

classifiers = [
    KNeighborsClassifier(3),
    SVC(kernel="rbf"),
    NuSVC(probability=True),
    DecisionTreeClassifier(),
    RandomForestClassifier(),
    AdaBoostClassifier(),
    GradientBoostingClassifier(),
    GaussianNB()]

# Logging for Visual Comparison
log_cols=["Classifier", "Accuracy"]
log = pd.DataFrame(columns=log_cols)

for clf in classifiers:
    clf.fit(X_train, y_train)
    name = clf.__class__.__name__    
    y_pred = clf.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    log_entry = pd.DataFrame([[name, acc*100]], columns=log_cols)
    log = log.append(log_entry)
````

````python
log=log.set_index('Classifier')
````

````python
plt.style.use('seaborn')
plt.barh(log.index,log['Accuracy'])
plt.xlabel('Accuracy')
plt.ylabel('Classifier')
plt.title('Accuracy of Various Classifier')
plt.show()
````

From the graph above we can see that Decision Tree Classifier, Support Vector Classifier and NuSVC model gave the best accuacy.
