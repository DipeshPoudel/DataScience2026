```python
import pandas as pd
```

```python
df = pd.read_csv("data_program.csv")
```

```python
df.head()
```

```python
df['LanguagesWorkedWith']
```

```python
lang_responses = df['LanguagesWorkedWith']
```

```python
from collections import Counter
```

```python
language_counter = Counter()
```

```python
for response in lang_responses:
    language_counter.update(response.split(';'))
```

```python
languages = []
popularity = []
```

```python
language_counter
```

```python
for item in language_counter.most_common(15):
    languages.append(item[0])
    popularity.append(item[1])
```

```python
languages
```

```python
plt.figure(figsize=(20,5))
plt.bar(languages,popularity,color='red')
plt.tight_layout()
```

```python
languages.reverse()
popularity.reverse()
```

```python
plt.barh(languages,popularity)
```

# Pie Chart

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/solar.csv")
```

```python
df.head()
```

```python
# explode=(0.1,0.1,0,0,0,0,0,0)
plt.figure(figsize=(20,5))
plt.pie(df['Installed Capacity (MW)'],
        labels=df['State'],
        autopct='%1.1f%%',
        explode=(0.1,0.1,0,0,0,0,0,0), 
        radius=1.5, 
        pctdistance=0.6,
        startangle=90,
        labeldistance=1,
        shadow=True)
plt.show()
```

Stacked and Grouped Bar Charts

```python
df.head()
```

```python
import numpy as np
```

```python
fig, ax = plt.subplots(figsize=(15,5))
width = 0.24
states = df['State']
x = np.arange(len(states))
generation = df['Generation (GWh)']
installed_capacity = df['Installed Capacity (MW)']
ax.bar((x-width/2),installed_capacity,width,label = 'Installed Capacity')
ax.bar((x+width/2),generation,width, label = 'Generation')
ax.set_xticks(x)
ax.set_xticklabels(states)
plt.legend()
plt.show()
```
