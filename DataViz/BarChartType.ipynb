```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data
x = ['USA', 'China', 'India', 'Germany', 'UK']
gold = [20, 15, 10, 8, 12]
silver = [25, 18, 12, 10, 14]
bronze = [15, 10, 8, 6, 9]

# Bar width
w = 0.25

# Position of bars on X-axis
bar1 = np.arange(len(x))
bar2 = [i + w for i in bar1]
bar3 = [i + w for i in bar2]

# Plotting the bars
plt.bar(bar1, gold, w, label='gold')
plt.bar(bar2, silver, w, label='silver')
plt.bar(bar3, bronze, w, label='bronze')

# Labels and title
plt.xlabel("Countries")
plt.ylabel("medals")
plt.title("Medals Vs Countries")
plt.xticks(bar1 + w, x)

plt.legend()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = {
    'Country': ['USA', 'China', 'India', 'Germany', 'UK'],
    'Gold': [20, 15, 10, 8, 12],
    'Silver': [25, 18, 12, 10, 14],
    'Bronze': [15, 10, 8, 6, 9]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Melt the DataFrame to tidy (long) format
df_melted = df.melt(id_vars='Country', 
                    value_vars=['Gold', 'Silver', 'Bronze'], 
                    var_name='Medal', 
                    value_name='Count')

# Plot using Seaborn
plt.figure(figsize=(8,5))
sns.barplot(x='Country', y='Count', hue='Medal', data=df_melted)

plt.title('Medals vs Countries')
plt.xlabel('Countries')
plt.ylabel('Medals')
plt.legend(title='Medal')
plt.show()
```
