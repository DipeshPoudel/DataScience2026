```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
```

```python
plt.figure(figsize=(20,5))
plt.plot(x,marker='o',linewidth=5,linestyle='dashed',markersize=12,color='#448191',label='Line 1')
plt.plot(y,marker='o',linewidth=5,linestyle='dashed',markersize=12,color='red',label='Line 2')
plt.xticks([0,1,2,3,4,5])
plt.yticks([2,4,6,8,10])
plt.title('A Very Simple Title')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.savefig('data_analysis_graph.png')
plt.legend(loc='lower right')
plt.show()
```
