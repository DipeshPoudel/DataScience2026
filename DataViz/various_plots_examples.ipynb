```python
fruits = ['apple','mango','guva']

online_sales = [5,6,8]
offline_sales = [6,7,8]
```

```python
bars1 = plt.bar(fruits,online_sales,label='Online Sales')
bars2 = plt.bar(fruits,offline_sales,bottom=online_sales,label='Offline Sales')
plt.bar_label(bars1,label_type="center")
plt.bar_label(bars2,label_type="center")
plt.legend(loc='upper left')
```

```python
import numpy as np
```

```python
x = np.arange(len(fruits))
width = 0.35
bars1 = plt.bar(x-width/2,online_sales,width=width,label='Online Sales')
bars2 = plt.bar(x+width/2,offline_sales,width=width, label = 'Offline Sales')
plt.bar_label(bars1,label_type="center")
plt.bar_label(bars2,label_type="center")
plt.xticks(x,fruits)
```

```python
x = [1,2,3,4,5]
fig,ax = plt.subplots(2,2,figsize=(10,5))
ax[0,0].plot(x,x)
ax[0,0].set_title('y=x')
ax[0,0].set_xlabel('Some Value')
ax[0,1].plot(x,[y**2 for y in x])
ax[0,1].set_title('y=x^2')
ax[1,0].scatter(x,x)
plt.tight_layout()
plt.show()
# print(ax)
```
