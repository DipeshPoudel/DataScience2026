```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('penguins')
```

```python
df.head()
```

### Histogram

```python
plt.hist(df['body_mass_g'])
plt.title('Histogram of Weight')
plt.show()
```

```python
plt.title('Histogram of Weight')
sns.histplot(data=df,x='body_mass_g',kde=True)
```

## Box Plot
