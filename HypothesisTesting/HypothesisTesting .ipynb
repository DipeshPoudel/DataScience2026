## Inferential Statistics

Inferential statistics is a branch of statistics that uses sample data to make generalizations, predictions, or inferences about a larger population. Unlike descriptive statistics, which summarize data, inferential statistics go beyond the data at hand to estimate parameters, test hypotheses, and predict future trends.

Key Characteristics of Inferential Statistics:
1. **Generalization**: Extends conclusions from a sample to a population.
2. **Hypothesis Testing**: Evaluates claims or predictions about a population.
3. **Estimation**: Provides parameter estimates (e.g., population mean, proportion) using sample statistics.
4. **Uncertainty Measurement**: Quantifies confidence and error margins.

**Example**: Using a survey of 500 people to estimate the voting preferences of an entire city’s population.

## Hypothesis in Statistics

In statistics, a **hypothesis** is a formal statement or assumption about a population parameter (e.g., the population mean or proportion). It is often based on prior knowledge, theory, or research, and it forms the basis of hypothesis testing.

There are two main types of hypotheses:

### Null Hypothesis ($H_0$):
- This is the default or initial assumption.
- It states that there is **no effect**, **no difference**, or **no relationship**.
- Example: $H_0: \mu = 50$ (The population mean is 50)

### Alternative Hypothesis ($H_1$ or $H_a$):
- This is what you want to prove or provide evidence for.
- It states that there is an effect, a difference, or a relationship.
- Example: $H_1: \mu ≠ 50$ (The population mean is not 50)

The goal of hypothesis testing is to see if there is enough evidence against the null hypothesis. In other words, to see if there is enough evidence to reject the null hypothesis. If there is not enough evidence, then we fail to reject the null hypothesis.

#### Example

A man, Mr. Orangejuice, goes to trial and is tried for the murder of his ex-wife. He is either guilty or innocent. Set up the null and 
alternative hypotheses for this example.

Putting this in a hypothesis testing framework, the hypotheses being tested are:
1. The man is guilty
2. The man is innocent

Let's set up the null and alternative hypotheses.

- $H_0$: Mr. Orangejuice is innocent
- $H_1$: Mr. Orangejuice is guilty

Remember that we assume the null hypothesis is true and try to see if we have evidence against the null. Therefore, it makes sense in this example to assume the man is innocent and test to see if there is evidence that he is guilty.

| Decision                           | Reality: $(H_0$) is true          | Reality: $(H_0$) is false         |
|------------------------------------|-----------------------------------|-----------------------------------|
| **Reject $(H_0$)** (conclude $(H_a$)) | Type I error                       | Correct decision                  |
| **Fail to reject $(H_0$)**         | Correct decision                  | Type II error                     |


The “reality”, or truth, about the null hypothesis is unknown and therefore we do not know if we have made the correct decision or if we committed an error. We can, however, define the likelihood of these events.

**Alpha $\alpha$**: The probability of committing a Type I error. Also known as the **significance level**.

**Beta $\beta$**: The probability of committing a Type II error.

**Power**: Power is the probability the null hypothesis is rejected given that it is false (ie. $1-\beta$)

$\alpha$ and $\beta$ are probabilities of committing an error so we want these values to be low. However, we cannot decrease both. As $\alpha$ decreases, $\beta$ increases.

## Steps for Hypothesis Testing



### Step 1: Set Up Hypotheses & Check Assumptions

- **Null hypothesis ($H_0$)**: assumption that the parameter equals a specific value (e.g. no effect).
- **Alternative hypothesis ($H_1$)**: opposite claim (e.g. effect, difference, or relationship).
- **Assumptions/Conditions**: e.g. normality, independence, sample size requirements 

### Step 2: Decide Significance Level (α)

- Typical choices: **0.05**, or sometimes **0.01** or **0.10**.
- Defines the threshold for “unlikely” outcomes under $H_0$ (Type I error rate) 

### Step 3: Compute Test Statistic

- Use appropriate statistic ($z, t, χ², F$) depending on context and assumptions.
- Formula (generic):  
  $$[
    \text{test stat} = \frac{\text{sample statistic} - \text{null parameter}}{\text{standard error}}
  ]$$ 

### Step 4: Find p‑Value or Critical Region

- **p‑value approach**: probability of observing a statistic as extreme or more assuming $H_0$ true.
- **Critical region approach**: compare test statistic to critical value from distribution given α :

### Step 5: Make a Decision

- If **$p ≤ α$** (or if test statistic falls in rejection region): **reject $H_0$**.
- Otherwise: **fail to reject H₀** (do not accept $H_0$ as true) 

### Step 6: State Conclusion

- Interpret results in plain language, relating back to research question.
- Example: “There is (or is not) sufficient evidence to conclude that …”

### Example 1: One-Sample t-test

You believe the average height of adult males is 175 cm.
You collect a sample of 10 individuals.

#### Hypotheses:
- $H_0: \mu = 175$ (population mean)
- $H_1: \mu ≠ 175$ (mean is different)

```python
import numpy as np
from scipy import stats

# Sample data
sample_heights = [172, 178, 169, 180, 174, 176, 177, 173, 171, 175]

# Perform one-sample t-test
t_stat, p_value = stats.ttest_1samp(sample_heights, popmean=175)

print("t-statistic:", t_stat)
print("p-value:", p_value)

# Decision
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

### Example 2: Two-Sample t-test (Independent Samples)

You're comparing test scores of students taught using Method A vs Method B.

#### Hypotheses:
- $H_0: \mu_1 = \mu_2$ (no difference)
- $H_0: \mu_1 \neq \mu_2$ (means difference)

```python
# Test scores
method_A = [78, 85, 82, 90, 88]
method_B = [72, 80, 75, 83, 79]

# Independent two-sample t-test
t_stat, p_value = stats.ttest_ind(method_A, method_B)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis: the methods differ")
else:
    print("Fail to reject the null hypothesis: no significant difference")
```

### Example 3: Proportions Test (Chi-Square Test for Independence)

You want to test if **gender and product preference** are independent.

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Contingency table (rows: gender, columns: product type)
data = pd.DataFrame([
    [30, 10],  # Male
    [20, 40]   # Female
], columns=["Product A", "Product B"], index=["Male", "Female"])
```

```python
chi2, p, dof, expected = chi2_contingency(data)

print("Chi-square stat:", chi2)
print("p-value:", p)

if p < 0.05:
    print("Reject the null: Gender and product preference are related")
else:
    print("Fail to reject the null: No significant relationship")
```

```python
import numpy as np
from scipy.stats import shapiro

# Case 1: Normally distributed data
normal_data = np.random.normal(loc=0, scale=1, size=100)

stat1, p_value1 = shapiro(normal_data)
print("Case 1: Normal Data")
print(f"Shapiro-Wilk Test Statistic = {stat1:.4f}, p-value = {p_value1:.4f}")
if p_value1 > 0.05:
    print("Fail to reject H0: Data looks normal\n")
else:
    print("Reject H0: Data does NOT look normal\n")


# Case 2: Non-normally distributed data (e.g., exponential)
non_normal_data = np.random.exponential(scale=1, size=100)

stat2, p_value2 = shapiro(non_normal_data)
print("Case 2: Non-Normal Data")
print(f"Shapiro-Wilk Test Statistic = {stat2:.4f}, p-value = {p_value2:.4f}")
if p_value2 > 0.05:
    print("Fail to reject H0: Data looks normal\n")
else:
    print("Reject H0: Data does NOT look normal\n")
```
