```python
import pandas as pd
import statsmodels.api as sm
```

```python
df = pd.read_csv("datasets/lifesat.csv")
```

```python
df.head()
```

```python
df = df.drop(columns=['Country'])
```

```python
df = df.rename(columns={
    'GDP per capita (USD)':'pci',
    'Life satisfaction': 'life_sat'
})
```

```python
df.head()
```

```python
X = df['pci']
X = sm.add_constant(X)  # Adds the intercept term
Y = df['life_sat']
```

```python
model = sm.OLS(Y, X)       # Ordinary Least Squares
results = model.fit()
```

```python
print(results.summary())
```
