```python
df = pd.read_csv("datasets/lifesat.csv")
```

```python
df.head()
```

```python
df = df.drop(columns=['Country'])
```

```python
df = df.rename(columns={
    "GDP per capita (USD)":"pci",
    "Life satisfaction":"life_sat"
})
```

```python
df.head()
```

```python
X = df['pci']
```

```python
X = sm.add_constant(X)
```

```python
Y = df['life_sat']
```

```python
model = sm.OLS(Y,X)
```

```python
result = model.fit()
```

```python
print(result.summary())
```
