```python
df = df.drop(columns=['Country'])
```

```python
df.head()
```

```python
df=df.rename(columns={'GDP per capita (USD)':'x','Life satisfaction':'y'})
```

```python
df.head()
```

```python
x_mean = df['x'].mean()
```

```python
x_mean
```

```python
y_mean = df['y'].mean()
```

```python
y_mean
```

```python
df['x_xmean'] = df['x']-x_mean
```

```python
df['y_ymean'] = df['y']-y_mean
```

```python
df['x_mean_sq'] = df['x_xmean']**2
```

```python
df.head()
```

```python
beta_1 = sum(df['x_xmean']*df['y_ymean'])/sum(df['x_mean_sq'])
```

```python
beta_1
```

```python
beta_0 = y_mean-beta_1*x_mean
```

```python
beta_0
```

```python
f'{beta_0}+{beta_1}x'
```
