```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Loading the Dataset

```python
df = pd.read_csv("datasets/lifesat.csv")
```

```python
df.head()
```

Removing the Column `Country`

```python
df = df.drop(columns=['Country'])
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns = ['per_capita_income','life_sat']
```

```python
df.head()
```

```python
mean_x = df['per_capita_income'].mean()
```

```python
mean_x
```

```python
mean_y = df['life_sat'].mean()
```

```python
mean_y
```

```python
df['x_diff_mean'] = df['per_capita_income']-mean_x
```

```python
df['y_diff_mean'] = df['life_sat']-mean_y
```

```python
df['x_diff_mean_2'] = df['x_diff_mean']**2
```

```python
df.head()
```

```python
beta_1 = sum(df['x_diff_mean']*df['y_diff_mean'])/sum(df['x_diff_mean_2'])
```

```python
beta_0 = mean_y - (beta_1*mean_x)
```

```python
print(f"The Equation of line is y={beta_0}+{beta_1}x")
```
