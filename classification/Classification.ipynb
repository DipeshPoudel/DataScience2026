# Classification 

## Basic Concept

**Google** defines word <i>classify</i> as `arrange (a group of people or things) in classes or categories according to shared qualities or characteristics`. In Machine Learning, Classification is the process to train the model which can predict the class(group) in the given information belongs. Classification is an example of `supervised learning`, meaning that the during the training process we need to provide data with labels.

**Examples**
1. A Loan officer in a bank have to decide whether the applicant is <strong>"Safe"</strong> or <strong>"risky"</strong>
2. A OCR(Optical Character Recognition) software to determine which is the given alphabet.

Example 1 is an example of binary classification where the predicted value could be one of two class. Example 2 is an example of multi class classification where the given input have to be classified in one of many classes.

Data classification is a two-step process, consisting of a learning step (where a classification model is constructed) and a classification step (where the model is used to predict class labels for given data).
