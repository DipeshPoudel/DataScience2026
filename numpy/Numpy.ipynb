### Dynamic Vs. Static Typing

In programming languages like C,C++,Java etc. we have to declare variable type explicitly but in dynamically typed programming language like python we do not need to specify the data type of the variable. Let us take and example of script that will add two numbers.

```c
#include<stdio.h>
void main(){
    int a=10;
    int b=20;
    int c;
    c=a+b;
    printf("The sum of %d and %d is %d",a,b,c);
}
```

This same program can be written in python as 

```python
a=10
b=20
c=a+b
print("sum of {0} and {1} is {2}".format(a,b,c))
```

In C, we must declare the data type of every variable (like `int a, b, c;`). In Python, we don’t need to declare types — we can just assign values directly. This makes Python programs quicker and easier to write.
However, it also affects **performance** and **memory usage**. Since Python doesn’t know the data type in advance, it can’t optimize the program as well as C. It also has to use more memory to store extra information (metadata) about each variable.

### Dimension of Array

An array can also contain array as its element. The number of array an variable holds can be considered as its dimension. 1-D array is just a ordered list of similar data. 2-D array is an array of array. 3-D array is an array that contains two arrarys as its element. A N-D array is an array that contains `N` number of arrays as its element. We can visulize a 1-D array as a row.

![1-D Array](images/1darray.PNG)

The figure above is a visulization is an 1 dimesional (1-D) array. A 2-D array can be visulized as a table. The figure below shows a `3X3` array. It means that each item in the array contains an array of 3.

![2-D Array](images/2darray.PNG)

The **number of dimensions** refers to how many axes an array has:
- A 1D array has one axis (like a simple list).
- A 2D array has two axes (rows and columns, like a matrix).
- A 3D array has three axes (depth, rows, columns).

We can check the number of dimensions using the `.ndim` attribute.

````python
arr1 = np.array([1, 2, 3])                  # 1D array
arr2 = np.array([[1, 2], [3, 4]])           # 2D array
arr3 = np.array([[[1], [2]], [[3], [4]]])   # 3D array

print("arr1 ndim:", arr1.ndim)
print("arr2 ndim:", arr2.ndim)
print("arr3 ndim:", arr3.ndim)
````

The **shape** of an array is a tuple that shows the number of elements along each dimension. For example:
- A shape of `(3,)` means the array has 3 elements in one dimension.
- A shape of `(2, 2)` means 2 rows and 2 columns.
- A shape of `(2, 2, 1)` means 2 blocks, each with 2 rows and 1 column.

We can check the shape using the `.shape` attribute.

````python
print("arr1 shape:", arr1.shape)
print("arr2 shape:", arr2.shape)
print("arr3 shape:", arr3.shape)
````

The **size** of an array is the **total number of elements** it contains. This is simply the product of the shape values.

For example, if the shape is `(2, 3)`, then the size is `2 × 3 = 6` elements.

We can check the total number of elements using the `.size` attribute.

````python
print("arr1 size:", arr1.size)
print("arr2 size:", arr2.size)
print("arr3 size:", arr3.size)
````

## Vectorization

Vectorization a style of computer programming where operations are applied to whole arrays instead of individual elements. Lets take an example using both pure python and numpy where we add each element of a array with corresponding element of same array.

````python
def element_wise_add_python(arr1,arr2):
    return [arr1+arr2 for (arr1,arr2) in zip(arr1,arr2)]
````

We can use numpy to perform vector operation or to perform vectorization.

````python
def element_wise_add_numpy(arr1,arr2):
    return np.add(arr1,arr2)
````

Lets use Simple benchmarking in a large array so that we can see the difference

````python
arr1 = np.random.randint(1,10000,50000)
arr2 = np.random.randint(50,10000,50000)
````

````python
%%timeit
X=element_wise_add_python(arr1,arr2)
````

````python
%%timeit
y=element_wise_add_numpy(arr1,arr2)
````

From the above result we can see how we can increase the performance by performing vectorization. Per operation the pure python method took around 16.3 micro second and the numpy took 31.9 nano second per operation which is much faster than pure python implementation.

## Cosine Simalirity Example 

The Angle between two vectors is given by
$$
\cos\theta = \frac{\vec{a}\cdot\vec{b}}{\|\vec{a}\|\;\|\vec{b}\|}
$$
![Angle Between Two Vectors](images/angle_between_vectors.png)

````python
import numpy as np
````

````python
# User-Movie rating matrix (rows = users, columns = movies)
ratings = np.array([
    [5, 4, 0, 1],  # User 1
    [4, 0, 0, 1],  # User 2
    [1, 1, 0, 5],  # User 3
    [0, 0, 5, 4]   # User 4
])
````

````python
25+16+1
````

````python
# Step 1: Compute dot product between all users
dot_product = ratings @ ratings.T  # shape (4,4)
````

````python
dot_product
````

````python
# Step 2: Compute norm (magnitude) of each user's vector
norms = np.linalg.norm(ratings, axis=1)
````

````python
norms
````

````python
# Step 4: Cosine similarity matrix
cosine_similarity = dot_product / (norms[:, None] * norms[None, :])
````

````python
print(np.round(cosine_similarity, 2))
````

This shows, for example, User 1 is most similar to User 2, less to User 3, and almost no similarity with User 4.
