# K-Nearest Neighbor (KNN)

K-Nearest Neighbor is a supervised, non-parametric, lazy machine learning algorithm that can be used for both `classification` and `regression`. 

* It is a non-parametric algorithm as it does not make any assumptions about the data distribution.
* It is a lazy algorithm as model is computed during classification and is dependent on the test set.

KNN algorithm is based on feature similarity meaning, the classification is carried out by finding the similarity metric (usually distance) between the data points. It uses k 'nearest' points for performing classification

## How Does it Work?

The K-NN algorithm is quite simple. We just need following things to build a KNN model.
1. Dataset with label data
2. Distance Metric
3. Value of k

### Distance Metric

The KNN algorithm is based on similarity which is calculated using distance between two points. We will discuss some of the popular ones.
1. Euclidean Distance: This is one of the famous one from `coordinate geometry`. It is the common formula to calculate distance between two points. 
\begin{align}
distance=\sqrt{\sum_{i=1}^k(x_{i}-y_{i})^2} \\
\end{align}
2. Manhattan Distance: This one takes the absolute value of difference between the points rather than square root like Euclidean distance.
\begin{align}
distance=\sum_{i=1}^k|(x_{i}-y_{i})| \\
\end{align}


To classify an unknown datapoint followings steps are taken
1. Compute distance to each datapoint
2. Identify k nearest neighbors
3. Use class labels of nearest neighbors to determine the class label of unknown record by taking the majority vote.

To understand this better let us suppose that we have two classes denoted in the figure below by `-` (minus) and `+` (plus) sign. If we need to classify new data point `X` then first of all we need to select k. For this example we have taken k=3. Now, 3 nearest point to `X` is selected. Among these three points 2 are in class plus and one is in minus. If we take majority vote of classes then the new point is classified as plus.

## Worked Out Example

We will build one classifier and one regressor using the KNN algorithm.

### Classification Using KNN

```python
import warnings
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
import numpy as np
warnings.filterwarnings('ignore')
```

Load the Dataset

```python
df = pd.read_csv("data/mushrooms.csv")
```

```python
df.head()
```

Check the Dataset and the Distribution of Target

```python
print("The total Number of Observations: {}".format(df.shape[0]))
print("Total Number of Edible Mushroom in the data set: {}".format(df['class'].value_counts()[0]))
print("Total Number of Poisonous Mushroom in the data set: {}".format(df['class'].value_counts()[1]))
```

Separate features (X) and target (y) from the data

```python
X = df.drop(columns=['class'],axis=1)
y = df['class']
```

Since all the features are categorical we need to convert them to numeric. One of the way is to create dummy variable. For example if we have the following dataset

| Feature1 | Feature1 | Feature3 |
|----------|----------|----------|
| a        | 1        | 2        |
| b        | 2        | 4        |
| a        | 5        | 9        |
| c        | 10       | 5        |

where Feature1 is categorical feature with values `a` , `b` and `c`. Then the feature1 is broken down into three features `feature1_a`,`feature1_b` and `feature1_c`. The value of each feature is either 1 or 0. If the value occurs in that row then respective feature 1 else 0. The dummy variables of this feature will be generated as 

| Feature1_a | Feature1_b | Feature1_c |
|------------|------------|------------|
| 1          | 0          | 0          |
| 0          | 1          | 0          |
| 1          | 0          | 0          |
| 0          | 0          | 1          |


```python
X_dummy = pd.get_dummies(X, prefix=X.columns, drop_first=True)
```

```python
X_dummy.head()
```

Set poisonous=0 and Edible=1

```python
y = y.map({'p':0,'e':1}).values
```

Splitting Data into Training and Testing Set

```python
X_train,X_test,y_train,y_test=train_test_split(X_dummy,y,test_size=0.3,random_state=0)
```

Building the Model

```python
clf = KNeighborsClassifier(n_neighbors=3)
```

Training the Model

```python
clf.fit(X_train,y_train)
```

Making the Prediction

```python
y_pred = clf.predict(X_test)
```

Calculate the Accuracy Score

```python
print("The Accuracy of the Model is {:3f}".format(metrics.accuracy_score(y_test,y_pred)))
```

## How to select the Best Value of K (n_neighbors) ?

We can use grid search technique to find the best value of k. We will build and see the accuracy score for each value of k from 1 to 25

```python
from sklearn.model_selection import GridSearchCV
K_values=[i for i in range(26)]
parameters = {"n_neighbors":K_values}
mode=GridSearchCV(clf, parameters,scoring='accuracy',verbose=1,refit=True,cv=5,n_jobs=-1)
```

```python
mode.fit(X_train,y_train)
```

```python
best_parameters = mode.best_params_
print(best_parameters)
```

The best value of k is 1 for this case

### Regression using KNN

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split
from sklearn import datasets
from sklearn import metrics
```

Load the Data

```python
boston = datasets.load_boston()
```

Get Feature (X) and Target (y)

```python
X=boston.data
y=boston.target
```

Split the dataset into training and testing set

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=0)
```

Build the Model

```python
reg=KNeighborsRegressor(n_neighbors=2)
```

Train the Model

```python
reg.fit(X_train,y_train)
```

 Make the Prediction

```python
y_pred=reg.predict(X_test)
```

Evaluate the Model

```python
# Different Metrics to Measure the Performance of the Model
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred))  
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print('R^2 (Coefficient of determination):', metrics.r2_score(y_test,y_pred))
```

Find the Best value of K for the Model

```python
from sklearn.model_selection import GridSearchCV
K_values=[i for i in range(26)]
parameters = {"n_neighbors":K_values}
mode=GridSearchCV(reg, parameters,scoring='r2',verbose=1,refit=True,cv=5,n_jobs=-1)
```

```python
mode.fit(X_train,y_train)
```

```python
best_parameters = mode.best_params_
print(best_parameters)
```

For the regression model we build the best value of k is 4
